# SchoolBridge 시장분석 — 단일 이미지 (HWP 삽입용)

## 수정 포인트
- **폰트**: `FONT_FAMILY` 변수
- **색상**: `COLOR_*` 변수
- **해상도**: `dpi=200`
- **텍스트**: `fig.text()` 내 문자열 직접 수정

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.gridspec as gridspec
import matplotlib.ticker as mticker
import numpy as np
import warnings
warnings.filterwarnings('ignore')

FONT_FAMILY = 'Malgun Gothic'
matplotlib.rcParams['font.family'] = FONT_FAMILY
matplotlib.rcParams['axes.unicode_minus'] = False

# ── 색상
C_WHITE = '#FFFFFF'
C_PANEL = '#F2F5F9'   # 영역 차트 패널 (매우 연한 회청)
C_TEXT  = '#1A202C'
C_GRAY  = '#4A5568'
C_LIGHT = '#CBD5E0'   # 스파인
C_NAVY  = '#1E3A5F'   # 막대 메인 색
C_NAVY2 = '#2C5282'   # 약간 밝은 네이비
C_NAVY3 = '#2B6CB0'   # 더 밝은 네이비

# 영역 차트 색
C_A1 = '#1E40AF'  # 국제결혼 (진한 파랑)
C_A2 = '#92400E'  # 중도입국 (짙은 황갈)
C_A3 = '#991B1B'  # 외국인가정 (짙은 빨강)

# ── 영역 차트 데이터
years = list(range(2012, 2026))
intermarriage = [29200, 32000, 40500, 50500, 60000, 67000, 75500, 83000,
                 83600, 86500, 80700, 79600, 79000, 75900]
midway        = [2600,  3000,  5500,  6700,  7200,  8500,  9500,  8700,
                 8700,  7900,  5800,  7200,  7200,  7200]
foreign       = [1800,  1500,  1400,  1600,  2000,  2500,  4500, 14000,
                 12000,  7000, 24000, 27900, 29300, 34000]
totals = [a + b + c for a, b, c in zip(intermarriage, midway, foreign)]

# ── 막대 차트 데이터 (경기·서울·인천 + 제주·세종만)
# y 위치: 0,1,2 → 경기·서울·인천 / y=3 gap / 4,5 → 제주·세종
bar_items = [
    ('경기', 35199, 30.2, 0),
    ('서울', 12563, 10.8, 1),
    ('인천',  9338,  8.0, 2),
    ('제주',  2127,  1.8, 4),
    ('세종',   538,  0.5, 5),
]

# ── Figure
fig = plt.figure(figsize=(26, 11), facecolor=C_WHITE)
gs = gridspec.GridSpec(1, 2, figure=fig,
                       width_ratios=[1.35, 1.0],
                       wspace=0.13,
                       left=0.06, right=0.97, top=0.88, bottom=0.12)

fig.suptitle('다문화 초등학생 현황 분석 (2012-2025)',
             fontsize=20, fontweight='bold', color=C_TEXT, y=0.97)
fig.text(0.5, 0.925, '출처: 한국교육개발원 교육통계서비스(KESS) 2025',
         ha='center', fontsize=9.5, color=C_GRAY)

# ══════════════════════════════════════════════════════
# 왼쪽: 스택 영역 차트
# ══════════════════════════════════════════════════════
ax1 = fig.add_subplot(gs[0])
ax1.set_facecolor(C_PANEL)

ax1.stackplot(np.array(years),
              np.array(intermarriage), np.array(midway), np.array(foreign),
              colors=[C_A1, C_A2, C_A3], alpha=0.86)

ax1.legend(handles=[
    mpatches.Patch(color=C_A1, alpha=0.86, label='국제결혼 가정-국내출생'),
    mpatches.Patch(color=C_A2, alpha=0.86, label='중도입국'),
    mpatches.Patch(color=C_A3, alpha=0.86, label='외국인 가정 (부모 둘 다 외국인)'),
], loc='upper left', fontsize=10, framealpha=0.92,
   facecolor=C_PANEL, edgecolor=C_LIGHT)

ax1.annotate('외국인 가정 비중\n2012: 5.3%  →  2025: 29.3%',
             xy=(2025, totals[-1]), xytext=(2018.5, 128000),
             fontsize=10.5, fontweight='bold', color='#7F1D1D',
             arrowprops=dict(arrowstyle='->', color='#7F1D1D', lw=1.8),
             bbox=dict(boxstyle='round,pad=0.4',
                       facecolor='#FEE2E2', edgecolor='#7F1D1D'))

ax1.scatter([2025], [totals[-1]], color='#C53030', s=80, zorder=5,
            edgecolors='#7F1D1D', linewidth=1.2)

ax1.set_xlim(2011.5, 2025.8)
ax1.set_ylim(0, 145000)
ax1.set_xlabel('연도', fontsize=11, color=C_GRAY)
ax1.set_ylabel('학생 수 (명)', fontsize=11, color=C_GRAY)
ax1.set_title('초등 다문화 학생 13년 추이\n3.5배 증가, 외국인 가정 비중 급증',
              fontsize=13, fontweight='bold', color=C_TEXT, pad=10)
ax1.yaxis.set_major_formatter(mticker.FuncFormatter(lambda v, _: f'{int(v):,}'))
ax1.set_xticks(years)
ax1.set_xticklabels([str(y) for y in years], fontsize=9,
                    rotation=45, ha='right')
ax1.tick_params(colors=C_GRAY)
for sp in ['top', 'right']:
    ax1.spines[sp].set_visible(False)
for sp in ['left', 'bottom']:
    ax1.spines[sp].set_color(C_LIGHT)

# ══════════════════════════════════════════════════════
# 오른쪽: 수평 막대 차트 (경기·서울·인천 + 제주·세종)
# ══════════════════════════════════════════════════════
ax2 = fig.add_subplot(gs[1])
ax2.set_facecolor(C_WHITE)

# 막대 색: 네이비 블루 계열
BAR_COLORS = {
    '경기': C_NAVY,
    '서울': C_NAVY,
    '인천': C_NAVY,
    '제주': C_NAVY2,
    '세종': C_NAVY3,
}

for name, val, ratio, yp in bar_items:
    bc = BAR_COLORS[name]
    ax2.barh(yp, val, color=bc, height=0.65,
             edgecolor='#93C5FD', linewidth=0.8)
    inside = val > 5000
    xp = val * 0.5 if inside else val + 280
    ha = 'center' if inside else 'left'
    lc = '#F0F9FF' if inside else C_TEXT
    ax2.text(xp, yp, f'{val:,}명  ({ratio}%)',
             va='center', ha=ha, fontsize=11.5,
             fontweight='bold', color=lc)

# 구분 표시 (인천↔제주 사이 생략)
ax2.text(-1200, 3, '⋯⋯  중간 12개 시도 생략  ⋯⋯',
         va='center', ha='left', fontsize=9.5, color=C_GRAY,
         style='italic')
ax2.axhline(y=3, xmin=0.04, xmax=0.95, color=C_LIGHT,
            linewidth=1.0, linestyle='--')

# y축 설정
ytick_pos   = [0, 1, 2, 4, 5]
ytick_label = ['경기', '서울', '인천', '제주', '세종']
ax2.set_yticks(ytick_pos)
ax2.set_yticklabels(ytick_label, fontsize=13, fontweight='bold', color=C_TEXT)
ax2.invert_yaxis()
ax2.set_ylim(5.55, -0.55)

ax2.set_xlabel('다문화 초등학생 수 (명)', fontsize=11, color=C_GRAY)
ax2.set_xlim(-2500, 43000)
ax2.xaxis.set_major_formatter(
    mticker.FuncFormatter(lambda v, _: f'{int(v):,}' if v >= 0 else ''))
ax2.tick_params(colors=C_GRAY)
for sp in ['top', 'right']:
    ax2.spines[sp].set_visible(False)
for sp in ['left', 'bottom']:
    ax2.spines[sp].set_color(C_LIGHT)

ax2.set_title('시도별 다문화 초등학생 분포 (2025)\n경기 30%, 수도권 49%',
              fontsize=13, fontweight='bold', color=C_TEXT, pad=10)

OUT = 'market_analysis_chart.png'
plt.savefig(OUT, dpi=200, bbox_inches='tight', facecolor=C_WHITE)
print('market_analysis_chart.png 저장 완료')
```
